# rasch-per quickstart

Core workflow: simulate responses, fit CTT and Rasch models, check dimensionality, and write an HTML report.

```python
from rasch_per import CTTAnalysis, RaschModel, ResponseData, generate_report
from rasch_per.dimensionality import run_pcar
from rasch_per.simulate import simulate_rasch_data
```

```python
# Synthetic data: 500 persons, 20 items, seeded for reproducibility
sim = simulate_rasch_data(n_persons=500, n_items=20, seed=42)
df = sim.set_index("person_id") if "person_id" in sim.columns else sim
df.head()
```

```python
data = ResponseData(df)

ctt = CTTAnalysis(data).run()
ctt.summary()
```

```python
model = RaschModel().fit(data, estimator="MML")
print("Item difficulties (logits):")
print(model.item_difficulties.round(3))
print("\nCronbach alpha:", round(ctt.reliability.cronbach_alpha, 3))
```

```python
pcar = run_pcar(model)
print("PCAR first-contrast eigenvalue:", round(pcar.first_contrast_eigenvalue, 3))
print("Second dimension suspected:", pcar.second_dimension_suspected)
```

```python
generate_report(df, output="quickstart_report.html")
print("Wrote quickstart_report.html")
```

Open `quickstart_report.html` in a browser. It includes CTT summaries, item difficulty / discrimination bars, a Wright map, the test-information curve, and the dimensionality check.
